```python

from pathlib import Path
from collections import Counter
import csv
import json

from sklearn.model_selection import train_test_split
from torchvision import datasets

# --------------------------------------------------
# 1. Paths: use the dataset location we verified
# --------------------------------------------------
DATASET_DIR = Path(
    "/content/drive/MyDrive/datasets/plant_disease/"
    "new_plant_disease_dataset/new_plant_disease_dataset"
)

TRAIN_DIR = DATASET_DIR / "train"
VALID_DIR = DATASET_DIR / "valid"

EXPERIMENT_DIR = Path(
    "/content/drive/MyDrive/PlantDisease_11Class_Experiment"
)
EXPERIMENT_DIR.mkdir(parents=True, exist_ok=True)

assert TRAIN_DIR.is_dir(), f"Missing: {TRAIN_DIR}"
assert VALID_DIR.is_dir(), f"Missing: {VALID_DIR}"

# --------------------------------------------------
# 2. Select our 11 target classes
# --------------------------------------------------
SELECTED_CLASSES = [
    "Corn_(maize)___Cercospora_leaf_spot Gray_leaf_spot",
    "Corn_(maize)___Common_rust_",
    "Corn_(maize)___Northern_Leaf_Blight",
    "Corn_(maize)___healthy",
    "Potato___Early_blight",
    "Potato___Late_blight",
    "Potato___healthy",
    "Tomato___Early_blight",
    "Tomato___Late_blight",
    "Tomato___Tomato_Yellow_Leaf_Curl_Virus",
    "Tomato___healthy",
]

# ImageFolder reads class folders and assigns numeric labels.
# We do not apply image transforms yet; this step checks paths and labels.
train_source = datasets.ImageFolder(str(TRAIN_DIR))
valid_source = datasets.ImageFolder(str(VALID_DIR))

assert train_source.classes == valid_source.classes, (
    "Train and validation class mappings differ. "
    "Inspect folder names before proceeding."
)

missing = set(SELECTED_CLASSES) - set(train_source.classes)
assert not missing, f"Selected classes missing from dataset: {missing}"

# --------------------------------------------------
# 3. Filter training samples to the selected classes
# --------------------------------------------------
class_to_label = {
    class_name: index
    for index, class_name in enumerate(SELECTED_CLASSES)
}

train_rows = [
    {
        "filepath": filepath,
        "class_name": class_name,
        "label": class_to_label[class_name],
        "split": "train",
    }
    for filepath, original_label in train_source.samples
    if (class_name := train_source.classes[original_label])
    in class_to_label
]

# --------------------------------------------------
# 4. Filter validation-source samples
# --------------------------------------------------
valid_rows = [
    {
        "filepath": filepath,
        "class_name": class_name,
        "label": class_to_label[class_name],
    }
    for filepath, original_label in valid_source.samples
    if (class_name := valid_source.classes[original_label])
    in class_to_label
]

assert train_rows, "No selected training images found."
assert valid_rows, "No selected validation images found."

# Every selected class must have enough validation images
source_valid_counts = Counter(row["class_name"] for row in valid_rows)

for class_name in SELECTED_CLASSES:
    count = source_valid_counts[class_name]
    assert count >= 2, (
        f"{class_name} has only {count} validation images; "
        "cannot create both validation and test sets."
    )

# --------------------------------------------------
# 5. Stratified split: 50% validation, 50% test
# --------------------------------------------------
valid_part, test_part = train_test_split(
    valid_rows,
    test_size=0.5,
    random_state=42,
    stratify=[row["label"] for row in valid_rows],
)

for row in valid_part:
    row["split"] = "validation"

for row in test_part:
    row["split"] = "test"

# --------------------------------------------------
# 6. Verify class representation in all three splits
# --------------------------------------------------
all_rows = train_rows + valid_part + test_part

split_counts = {
    split: Counter(
        row["class_name"] for row in all_rows
        if row["split"] == split
    )
    for split in ("train", "validation", "test")
}

print(f"{'CLASS':59} {'TRAIN':>7} {'VALID':>7} {'TEST':>7}")
print("-" * 84)

for class_name in SELECTED_CLASSES:
    counts = [
        split_counts[split][class_name]
        for split in ("train", "validation", "test")
    ]

    print(
        f"{class_name[:58]:59} "
        f"{counts[0]:7} {counts[1]:7} {counts[2]:7}"
    )

print("-" * 84)

for split in ("train", "validation", "test"):
    counts = split_counts[split]
    total = sum(counts.values())
    absent = [c for c in SELECTED_CLASSES if counts[c] == 0]

    print(
        f"{split.upper():12} total={total:6} | "
        f"classes represented={sum(counts[c] > 0 for c in SELECTED_CLASSES)}/"
        f"{len(SELECTED_CLASSES)}"
    )

    assert not absent, f"Classes missing from {split}: {absent}"

# --------------------------------------------------
# 7. Save manifest and class mapping
# --------------------------------------------------
manifest_path = EXPERIMENT_DIR / "data_manifest.csv"

with open(manifest_path, "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(
        f,
        fieldnames=["filepath", "class_name", "label", "split"],
    )
    writer.writeheader()
    writer.writerows(all_rows)

mapping_path = EXPERIMENT_DIR / "class_mapping.json"

with open(mapping_path, "w", encoding="utf-8") as f:
    json.dump(
        {str(i): name for i, name in enumerate(SELECTED_CLASSES)},
        f,
        indent=2,
    )

print("\nSaved manifest:", manifest_path)
print("Saved class mapping:", mapping_path)
print("Random seed: 42")
```

```python
import os
import json
import pandas as pd
from PIL import Image

import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

# --------------------------------------------------
# 1. Configure paths and reproducibility
# --------------------------------------------------
EXPERIMENT_DIR = (
    "/content/drive/MyDrive/PlantDisease_11Class_Experiment"
)
MANIFEST_PATH = os.path.join(EXPERIMENT_DIR, "data_manifest.csv")
CLASS_MAPPING_PATH = os.path.join(EXPERIMENT_DIR, "class_mapping.json")

SEED = 42
BATCH_SIZE = 32
IMAGE_SIZE = 224
NUM_WORKERS = 2

torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

assert os.path.isfile(MANIFEST_PATH), "Manifest CSV was not found."
assert os.path.isfile(CLASS_MAPPING_PATH), "Class mapping JSON was not found."

# --------------------------------------------------
# 2. Read and validate the saved manifest
# --------------------------------------------------
df = pd.read_csv(MANIFEST_PATH)

required_columns = {"filepath", "class_name", "label", "split"}
assert required_columns.issubset(df.columns), (
    f"Manifest must contain these columns: {required_columns}"
)

df["filepath"] = df["filepath"].astype(str)
df["class_name"] = df["class_name"].astype(str)
df["label"] = df["label"].astype(int)
df["split"] = df["split"].astype(str).str.lower().str.strip()

# Allow common names for the validation split.
df["split"] = df["split"].replace({
    "valid": "validation",
    "val": "validation"
})

expected_splits = {"train", "validation", "test"}
assert set(df["split"].unique()) == expected_splits, (
    f"Unexpected split names: {sorted(df['split'].unique())}"
)

# Rebuild the label mapping from the manifest itself.
class_pairs = df[["class_name", "label"]].drop_duplicates()

assert not class_pairs["class_name"].duplicated().any(), (
    "A class name has more than one numeric label."
)
assert not class_pairs["label"].duplicated().any(), (
    "A numeric label is assigned to more than one class."
)

class_to_idx = dict(
    zip(class_pairs["class_name"], class_pairs["label"])
)
idx_to_class = {
    label: class_name
    for class_name, label in class_to_idx.items()
}

assert len(class_to_idx) == 11, (
    f"Expected 11 classes, found {len(class_to_idx)}."
)
assert sorted(idx_to_class.keys()) == list(range(11)), (
    "Labels must be consecutive integers from 0 to 10."
)

# Verify that each split contains every class.
for split_name, split_df in df.groupby("split"):
    assert set(split_df["label"].unique()) == set(range(11)), (
        f"Not all classes are present in {split_name}."
    )

print("\nImages per split:")
print(df.groupby("split").size().to_string())
print("\nNumber of classes:", len(class_to_idx))

# --------------------------------------------------
# 3. Define image transformations
# --------------------------------------------------
imagenet_mean = (0.485, 0.456, 0.406)
imagenet_std = (0.229, 0.224, 0.225)

train_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=10),
    transforms.ToTensor(),
    transforms.Normalize(imagenet_mean, imagenet_std),
])

evaluation_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(imagenet_mean, imagenet_std),
])

# --------------------------------------------------
# 4. Define how an image and label are loaded
# --------------------------------------------------
class PlantDiseaseDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, index):
        row = self.dataframe.iloc[index]

        image_path = row["filepath"]
        label = int(row["label"])

        with Image.open(image_path) as image:
            image = image.convert("RGB")

        if self.transform is not None:
            image = self.transform(image)

        return image, label

# --------------------------------------------------
# 5. Build the three datasets
# --------------------------------------------------
train_df = df[df["split"] == "train"].copy()
valid_df = df[df["split"] == "validation"].copy()
test_df = df[df["split"] == "test"].copy()

train_dataset = PlantDiseaseDataset(train_df, train_transform)
valid_dataset = PlantDiseaseDataset(valid_df, evaluation_transform)
test_dataset = PlantDiseaseDataset(test_df, evaluation_transform)

# --------------------------------------------------
# 6. Build batch loaders
# --------------------------------------------------
loader_options = {
    "batch_size": BATCH_SIZE,
    "num_workers": NUM_WORKERS,
    "pin_memory": (device.type == "cuda"),
}

train_loader = DataLoader(
    train_dataset,
    shuffle=True,
    **loader_options
)

valid_loader = DataLoader(
    valid_dataset,
    shuffle=False,
    **loader_options
)

test_loader = DataLoader(
    test_dataset,
    shuffle=False,
    **loader_options
)

# --------------------------------------------------
# 7. Test one batch before training
# --------------------------------------------------
images, labels = next(iter(train_loader))

print("\nFirst training batch:")
print("Image tensor shape:", tuple(images.shape))
print("Label tensor shape:", tuple(labels.shape))
print("Image data type:", images.dtype)
print("Label range:", int(labels.min()), "to", int(labels.max()))
print("\nData loaders are ready.")
```

```python
# --------------------------------------------------
# 8. Create the ResNet-18 model
# --------------------------------------------------

import torch.nn as nn
from torchvision import models

# Number of disease classes we want to predict
NUM_CLASSES = len(class_to_idx)

# Load ResNet-18 with pretrained ImageNet weights
model = models.resnet18(
    weights=models.ResNet18_Weights.IMAGENET1K_V1
)

# Freeze the pretrained layers for the first training stage
for param in model.parameters():
    param.requires_grad = False

# Find the number of inputs to the final classification layer
num_features = model.fc.in_features

# Replace the original classification layer with our 11-class layer
model.fc = nn.Linear(num_features, NUM_CLASSES)

# Move the model to the GPU when available
model = model.to(device)

# Check that the model is configured correctly
print("Model:", model.__class__.__name__)
print("Number of classes:", NUM_CLASSES)
print("Input features to final layer:", num_features)
print("Output features from final layer:", model.fc.out_features)

print(
    "Trainable parameters:",
    sum(p.numel() for p in model.parameters() if p.requires_grad)
)

print(
    "Total parameters:",
    sum(p.numel() for p in model.parameters())
)
```

```python
# --------------------------------------------------
# 9. Define the loss function and optimizer
# --------------------------------------------------

import torch.nn as nn
import torch.optim as optim

# 1. Define the loss function
criterion = nn.CrossEntropyLoss()

# 2. Define the optimizer
optimizer = optim.Adam(
    model.fc.parameters(),
    lr=0.001
)

print("Criterion:", criterion)
print("Optimizer:", optimizer.__class__.__name__)
print("Learning rate:", optimizer.param_groups[0]["lr"])
print(
    "Parameters being optimized:",
    sum(
        p.numel()
        for group in optimizer.param_groups
        for p in group["params"]
    )
)
```

```python

# --------------------------------------------------
# 10. Train and validate the model
# --------------------------------------------------

import os
import copy
import torch

# Save this experiment separately from previous experiments.
RUN_DIR = os.path.join(EXPERIMENT_DIR, "resnet18_run_01")
os.makedirs(RUN_DIR, exist_ok=True)

# Store metrics from every epoch.
history = []

# Keep track of the best validation result.
best_val_loss = float("inf")
best_model_state = copy.deepcopy(model.state_dict())

num_epochs = 1

for epoch in range(num_epochs):

    print(f"\nEpoch {epoch + 1}/{num_epochs}")
    print("-" * 40)

    # Run training first, then validation.
    for phase in ["train", "valid"]:

        if phase == "train":
            model.train()
            dataloader = train_loader
        else:
            model.eval()
            dataloader = valid_loader

        running_loss = 0.0
        running_corrects = 0
        total_images = 0

        for inputs, labels in dataloader:

            inputs = inputs.to(device)
            labels = labels.to(device)

            # Reset gradients for this training batch.
            if phase == "train":
                optimizer.zero_grad()

            # Disable gradient calculations during validation.
            with torch.set_grad_enabled(phase == "train"):

                # Forward pass: produce 11 logits per image.
                outputs = model(inputs)

                # Identify the predicted class.
                _, preds = torch.max(outputs, 1)

                # Compare predictions with the true labels.
                loss = criterion(outputs, labels)

                # Backpropagation and parameter updates
                # happen only during training.
                if phase == "train":
                    loss.backward()
                    optimizer.step()

            # Accumulate loss and correct predictions.
            batch_size = inputs.size(0)

            running_loss += loss.item() * batch_size
            running_corrects += (preds == labels).sum().item()
            total_images += batch_size

        # Calculate metrics across the entire dataset.
        epoch_loss = running_loss / total_images
        epoch_acc = running_corrects / total_images

        print(
            f"{phase.capitalize()} Loss: {epoch_loss:.4f} | "
            f"{phase.capitalize()} Accuracy: {epoch_acc:.4f} "
            f"({epoch_acc * 100:.2f}%)"
        )

        # Save the metrics for later analysis.
        history.append({
            "epoch": epoch + 1,
            "phase": phase,
            "loss": epoch_loss,
            "accuracy": epoch_acc
        })

        # Keep the model with the lowest validation loss.
        if phase == "valid" and epoch_loss < best_val_loss:
            best_val_loss = epoch_loss
            best_model_state = copy.deepcopy(model.state_dict())

            torch.save(
                {
                    "epoch": epoch + 1,
                    "model_state_dict": best_model_state,
                    "class_to_idx": class_to_idx,
                    "idx_to_class": idx_to_class,
                    "best_val_loss": best_val_loss
                },
                os.path.join(RUN_DIR, "best_model.pth")
            )

# Save the recorded metrics.
import pandas as pd

history_df = pd.DataFrame(history)
history_df.to_csv(
    os.path.join(RUN_DIR, "training_history.csv"),
    index=False
)

print("\nInitial training run completed.")
print("Best validation loss:", round(best_val_loss, 4))
print("Experiment directory:", RUN_DIR)
```

```python

# --------------------------------------------------
# 11. Reusable training and validation function
# --------------------------------------------------

import copy
import time
import pandas as pd
import torch
import torch.nn as nn
from tqdm.auto import tqdm


def train_model(
    model,
    criterion,
    optimizer,
    train_loader,
    valid_loader,
    stage_name,
    max_epochs,
    patience,
    min_delta,
    run_dir,
    history,
    use_amp=True
):
    best_val_loss = float("inf")
    best_model_state = copy.deepcopy(model.state_dict())
    epochs_without_improvement = 0

    checkpoint_path = os.path.join(
        run_dir, f"best_{stage_name}_model.pth"
    )
    history_path = os.path.join(
        run_dir, "training_history.csv"
    )

    amp_enabled = use_amp and torch.cuda.is_available()
    scaler = torch.amp.GradScaler(
        "cuda", enabled=amp_enabled
    )

    for epoch in range(max_epochs):
        print(
            f"\n{stage_name} | "
            f"Epoch {epoch + 1}/{max_epochs}"
        )

        for phase in ["train", "valid"]:
            model.train() if phase == "train" else model.eval()

            loader = (
                train_loader if phase == "train"
                else valid_loader
            )

            running_loss = 0.0
            running_corrects = 0
            total_images = 0
            start_time = time.time()

            progress = tqdm(
                loader,
                desc=f"{phase.capitalize()}",
                leave=True
            )

            for inputs, labels in progress:
                inputs = inputs.to(
                    device, non_blocking=True
                )
                labels = labels.to(
                    device, non_blocking=True
                )

                if phase == "train":
                    optimizer.zero_grad(set_to_none=True)

                with torch.set_grad_enabled(phase == "train"):
                    with torch.autocast(
                        device_type="cuda",
                        dtype=torch.float16,
                        enabled=amp_enabled
                    ):
                        outputs = model(inputs)
                        loss = criterion(outputs, labels)

                    if phase == "train":
                        scaler.scale(loss).backward()
                        scaler.step(optimizer)
                        scaler.update()

                batch_size = labels.size(0)
                running_loss += loss.detach().item() * batch_size
                running_corrects += (
                    outputs.detach().argmax(dim=1) == labels
                ).sum().item()
                total_images += batch_size

                progress.set_postfix(
                    loss=f"{running_loss / total_images:.4f}",
                    acc=f"{running_corrects / total_images:.3%}"
                )

            epoch_loss = running_loss / total_images
            epoch_acc = running_corrects / total_images
            elapsed = time.time() - start_time

            print(
                f"{phase.capitalize()} | "
                f"Loss: {epoch_loss:.4f} | "
                f"Accuracy: {epoch_acc:.2%} | "
                f"Time: {elapsed / 60:.2f} min"
            )

            history.append({
                "stage": stage_name,
                "epoch": epoch + 1,
                "phase": phase,
                "loss": epoch_loss,
                "accuracy": epoch_acc,
                "elapsed_seconds": elapsed
            })

            # Persist history after every completed phase.
            pd.DataFrame(history).to_csv(
                history_path, index=False
            )

            if phase == "valid":
                if epoch_loss < best_val_loss - min_delta:
                    best_val_loss = epoch_loss
                    best_model_state = copy.deepcopy(
                        model.state_dict()
                    )
                    epochs_without_improvement = 0

                    torch.save(
                        {
                            "stage": stage_name,
                            "epoch": epoch + 1,
                            "model_state_dict": best_model_state,
                            "optimizer_state_dict": optimizer.state_dict(),
                            "class_to_idx": class_to_idx,
                            "idx_to_class": idx_to_class,
                            "best_val_loss": best_val_loss
                        },
                        checkpoint_path
                    )

                    print("Best checkpoint saved.")
                else:
                    epochs_without_improvement += 1
                    print(
                        "No meaningful validation-loss improvement: "
                        f"{epochs_without_improvement}/{patience}"
                    )

        if epochs_without_improvement >= patience:
            print(f"Early stopping: {stage_name}")
            break

    # Restore the best weights found during this stage.
    model.load_state_dict(best_model_state)

    print(f"\nBest {stage_name} validation loss: {best_val_loss:.4f}")
    print(f"Checkpoint: {checkpoint_path}")

    return model, history
```

```python
import shutil
from pathlib import Path
from concurrent.futures import ThreadPoolExecutor, as_completed
from tqdm.auto import tqdm
import pandas as pd

SOURCE_ROOT = Path(
    "/content/drive/MyDrive/datasets/plant_disease/"
    "new_plant_disease_dataset/new_plant_disease_dataset"
)
LOCAL_ROOT = Path("/content/plant_disease_11class")
MANIFEST_PATH = Path(
    "/content/drive/MyDrive/PlantDisease_11Class_Experiment/"
    "data_manifest.csv"
)

SELECTED_CLASSES = [
    "Corn_(maize)___Cercospora_leaf_spot Gray_leaf_spot",
    "Corn_(maize)___Common_rust_",
    "Corn_(maize)___Northern_Leaf_Blight",
    "Corn_(maize)___healthy",
    "Potato___Early_blight",
    "Potato___Late_blight",
    "Potato___healthy",
    "Tomato___Early_blight",
    "Tomato___Late_blight",
    "Tomato___Tomato_Yellow_Leaf_Curl_Virus",
    "Tomato___healthy",
]

assert SOURCE_ROOT.exists(), f"Dataset root not found: {SOURCE_ROOT}"
assert MANIFEST_PATH.is_file(), f"Manifest not found: {MANIFEST_PATH}"

df = pd.read_csv(MANIFEST_PATH)
df["split"] = (
    df["split"].astype(str).str.strip().str.lower()
    .replace({"val": "validation", "valid": "validation"})
)
df = df[
    df["class_name"].isin(SELECTED_CLASSES)
    & df["split"].isin(["train", "validation", "test"])
].copy()

assert len(df) == 25_845, f"Unexpected manifest row count: {len(df):,}"

def resolve_source(value):
    p = Path(str(value))

    if p.is_file():
        return p.resolve()

    candidate = SOURCE_ROOT / p
    if candidate.is_file():
        return candidate.resolve()

    positions = [
        i for i, part in enumerate(p.parts)
        if part in ("train", "valid")
    ]
    for i in reversed(positions):
        candidate = SOURCE_ROOT / Path(*p.parts[i:])
        if candidate.is_file():
            return candidate.resolve()

    raise FileNotFoundError(f"Cannot locate source image: {value}")

def destination_for(src):
    try:
        relative = src.relative_to(SOURCE_ROOT)
    except ValueError:
        positions = [
            i for i, part in enumerate(src.parts)
            if part in ("train", "valid")
        ]
        if not positions:
            raise ValueError(f"Cannot determine relative path: {src}")
        relative = Path(*src.parts[positions[-1]:])
    return LOCAL_ROOT / relative

# Stage 1: show progress while resolving manifest paths.
unique_values = df["filepath"].drop_duplicates().tolist()
jobs_by_source = {}

with tqdm(total=len(unique_values), desc="Preparing paths", unit="image") as bar:
    for value in unique_values:
        src = resolve_source(value)
        jobs_by_source[str(src)] = (src, destination_for(src))
        bar.update(1)

jobs = list(jobs_by_source.values())
LOCAL_ROOT.mkdir(parents=True, exist_ok=True)

def copy_one(job):
    src, dst = job
    dst.parent.mkdir(parents=True, exist_ok=True)

    if dst.is_file() and dst.stat().st_size == src.stat().st_size:
        return "reused"

    shutil.copy2(src, dst)

    if not dst.is_file() or dst.stat().st_size != src.stat().st_size:
        raise IOError(f"Verification failed: {dst}")

    return "copied"

# Stage 2: live progress while copying.
copied = 0
reused = 0
errors = []

with ThreadPoolExecutor(max_workers=8) as pool:
    futures = [pool.submit(copy_one, job) for job in jobs]

    with tqdm(total=len(futures), desc="Copying images", unit="image") as bar:
        for future in as_completed(futures):
            try:
                result = future.result()
                if result == "copied":
                    copied += 1
                else:
                    reused += 1
            except Exception as exc:
                errors.append(str(exc))
            finally:
                bar.update(1)
                bar.set_postfix(
                    copied=copied,
                    reused=reused,
                    errors=len(errors),
                )

if errors:
    raise RuntimeError(
        f"{len(errors)} image operations failed. First errors:\n"
        + "\n".join(errors[:5])
    )

# Save manifest with local image paths.
def local_path(value):
    return str(destination_for(resolve_source(value)))

df["filepath"] = df["filepath"].map(local_path)

missing = [p for p in df["filepath"] if not Path(p).is_file()]
assert not missing, f"{len(missing)} manifest images are missing."

local_manifest = LOCAL_ROOT / "data_manifest_local.csv"
df.to_csv(local_manifest, index=False)

print("\nCOPY AND VERIFICATION COMPLETE")
print(f"Manifest rows: {len(df):,}")
print(f"Newly copied: {copied:,}")
print(f"Existing images reused: {reused:,}")
print(f"Local manifest: {local_manifest}")
```

```python
import os
import copy
import json
import time
import pandas as pd
import torch
from tqdm.auto import tqdm

STAGE1_EPOCHS = 10
PATIENCE = 3
MIN_DELTA = 0.001

best_val_loss = float("inf")
best_epoch = 0
epochs_without_improvement = 0
history = []

best_checkpoint_path = os.path.join(
    CHECKPOINT_DIR, "stage1_best.pt"
)
last_checkpoint_path = os.path.join(
    CHECKPOINT_DIR, "stage1_last.pt"
)
history_path = os.path.join(RUN_DIR, "stage1_history.csv")

# Reset the optimizer's gradients after the pre-flight test.
optimizer.zero_grad(set_to_none=True)

print(f"Starting Stage 1 on {device}")
print(f"Training images: {len(train_dataset):,}")
print(f"Validation images: {len(valid_dataset):,}")
print(f"Maximum epochs: {STAGE1_EPOCHS}")
print(f"Checkpoint directory: {CHECKPOINT_DIR}")

for epoch in range(1, STAGE1_EPOCHS + 1):
    epoch_start = time.time()

    # -------------------------
    # Training
    # -------------------------
    model.train()
    train_loss_sum = 0.0
    train_correct = 0
    train_count = 0

    train_bar = tqdm(
        train_loader,
        desc=f"Epoch {epoch}/{STAGE1_EPOCHS} [Train]",
        leave=True,
    )

    for images, labels in train_bar:
        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        optimizer.zero_grad(set_to_none=True)

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        batch_size = labels.size(0)
        train_loss_sum += loss.item() * batch_size
        train_correct += (outputs.argmax(dim=1) == labels).sum().item()
        train_count += batch_size

        train_bar.set_postfix(
            loss=f"{loss.item():.4f}",
            accuracy=f"{100 * train_correct / train_count:.2f}%",
        )

    train_loss = train_loss_sum / train_count
    train_accuracy = train_correct / train_count

    # -------------------------
    # Validation
    # -------------------------
    model.eval()
    val_loss_sum = 0.0
    val_correct = 0
    val_count = 0

    with torch.inference_mode():
        val_bar = tqdm(
            valid_loader,
            desc=f"Epoch {epoch}/{STAGE1_EPOCHS} [Validation]",
            leave=True,
        )

        for images, labels in val_bar:
            images = images.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            outputs = model(images)
            loss = criterion(outputs, labels)

            batch_size = labels.size(0)
            val_loss_sum += loss.item() * batch_size
            val_correct += (
                outputs.argmax(dim=1) == labels
            ).sum().item()
            val_count += batch_size

            val_bar.set_postfix(
                loss=f"{loss.item():.4f}",
                accuracy=f"{100 * val_correct / val_count:.2f}%",
            )

    val_loss = val_loss_sum / val_count
    val_accuracy = val_correct / val_count
    epoch_seconds = time.time() - epoch_start

    metrics = {
        "epoch": epoch,
        "train_loss": train_loss,
        "train_accuracy": train_accuracy,
        "val_loss": val_loss,
        "val_accuracy": val_accuracy,
        "epoch_seconds": epoch_seconds,
    }
    history.append(metrics)

    # Save history after every epoch, so progress is preserved.
    pd.DataFrame(history).to_csv(history_path, index=False)

    # Save the latest model state.
    torch.save(
        {
            "epoch": epoch,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "metrics": metrics,
        },
        last_checkpoint_path,
    )

    # Early stopping monitors validation loss.
    if val_loss < best_val_loss - MIN_DELTA:
        best_val_loss = val_loss
        best_epoch = epoch
        epochs_without_improvement = 0

        torch.save(
            {
                "epoch": epoch,
                "model_state_dict": copy.deepcopy(model.state_dict()),
                "optimizer_state_dict": optimizer.state_dict(),
                "metrics": metrics,
                "best_val_loss": best_val_loss,
            },
            best_checkpoint_path,
        )
        status = "Best checkpoint saved"
    else:
        epochs_without_improvement += 1
        status = (
            f"No qualifying improvement "
            f"({epochs_without_improvement}/{PATIENCE})"
        )

    print(
        f"\nEpoch {epoch}/{STAGE1_EPOCHS} | "
        f"Train loss: {train_loss:.4f} | "
        f"Train accuracy: {train_accuracy:.2%} | "
        f"Val loss: {val_loss:.4f} | "
        f"Val accuracy: {val_accuracy:.2%} | "
        f"Time: {epoch_seconds:.1f}s | {status}\n"
    )

    if epochs_without_improvement >= PATIENCE:
        print("Early stopping: validation loss has stopped improving.")
        break

# Restore the best validation-loss checkpoint for the next stage.
checkpoint = torch.load(
    best_checkpoint_path,
    map_location=device,
    weights_only=True,
)
model.load_state_dict(checkpoint["model_state_dict"])

print("\nSTAGE 1 COMPLETE")
print(f"Best epoch: {best_epoch}")
print(f"Best validation loss: {best_val_loss:.4f}")
print(f"Best checkpoint: {best_checkpoint_path}")
print(f"Training history: {history_path}")
```

```python

import os
import copy
import time
import pandas as pd
import torch
import torch.nn as nn
from tqdm.auto import tqdm

# Stage 2: fine-tune the final ResNet block and classifier
STAGE2_EPOCHS = 10
STAGE2_PATIENCE = 3
STAGE2_MIN_DELTA = 0.001

# Freeze the entire network first
for param in model.parameters():
    param.requires_grad = False

# Unfreeze only the final convolutional block and classifier
for param in model.layer4.parameters():
    param.requires_grad = True

for param in model.fc.parameters():
    param.requires_grad = True

# Smaller learning rate for pretrained features;
# larger learning rate for the new classification layer
optimizer = torch.optim.Adam(
    [
        {"params": model.layer4.parameters(), "lr": 1e-4},
        {"params": model.fc.parameters(), "lr": 1e-3},
    ]
)

best_val_loss = float("inf")
best_epoch = 0
epochs_without_improvement = 0
stage2_history = []

stage2_best_path = os.path.join(
    CHECKPOINT_DIR, "stage2_best.pt"
)
stage2_last_path = os.path.join(
    CHECKPOINT_DIR, "stage2_last.pt"
)
stage2_history_path = os.path.join(
    RUN_DIR, "stage2_history.csv"
)

print("Starting Stage 2 fine-tuning")
print("Trainable blocks: layer4 and fc")
print("Earlier ResNet blocks remain frozen.")
print(f"Training images: {len(train_dataset):,}")
print(f"Validation images: {len(valid_dataset):,}")

for epoch in range(1, STAGE2_EPOCHS + 1):
    start_time = time.time()

    model.train()

    # Keep BatchNorm statistics fixed in frozen blocks.
    # BatchNorm in trainable blocks can still adapt.
    for module in model.modules():
        if isinstance(module, nn.BatchNorm2d):
            if not any(
                p.requires_grad
                for p in module.parameters(recurse=False)
            ):
                module.eval()

    train_loss_sum = 0.0
    train_correct = 0
    train_count = 0

    train_bar = tqdm(
        train_loader,
        desc=f"Stage 2 Epoch {epoch}/{STAGE2_EPOCHS} [Train]"
    )

    for images, labels in train_bar:
        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        optimizer.zero_grad(set_to_none=True)
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        n = labels.size(0)
        train_loss_sum += loss.item() * n
        train_correct += (
            outputs.argmax(dim=1) == labels
        ).sum().item()
        train_count += n

        train_bar.set_postfix(
            loss=f"{loss.item():.4f}",
            accuracy=f"{100 * train_correct / train_count:.2f}%"
        )

    train_loss = train_loss_sum / train_count
    train_accuracy = train_correct / train_count

    model.eval()
    val_loss_sum = 0.0
    val_correct = 0
    val_count = 0

    with torch.inference_mode():
        val_bar = tqdm(
            valid_loader,
            desc=f"Stage 2 Epoch {epoch}/{STAGE2_EPOCHS} [Validation]"
        )

        for images, labels in val_bar:
            images = images.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            outputs = model(images)
            loss = criterion(outputs, labels)

            n = labels.size(0)
            val_loss_sum += loss.item() * n
            val_correct += (
                outputs.argmax(dim=1) == labels
            ).sum().item()
            val_count += n

            val_bar.set_postfix(
                loss=f"{loss.item():.4f}",
                accuracy=f"{100 * val_correct / val_count:.2f}%"
            )

    val_loss = val_loss_sum / val_count
    val_accuracy = val_correct / val_count
    elapsed = time.time() - start_time

    metrics = {
        "epoch": epoch,
        "train_loss": train_loss,
        "train_accuracy": train_accuracy,
        "val_loss": val_loss,
        "val_accuracy": val_accuracy,
        "epoch_seconds": elapsed,
    }

    stage2_history.append(metrics)
    pd.DataFrame(stage2_history).to_csv(
        stage2_history_path, index=False
    )

    torch.save(
        {
            "epoch": epoch,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "metrics": metrics,
        },
        stage2_last_path,
    )

    if val_loss < best_val_loss - STAGE2_MIN_DELTA:
        best_val_loss = val_loss
        best_epoch = epoch
        epochs_without_improvement = 0

        torch.save(
            {
                "epoch": epoch,
                "model_state_dict": copy.deepcopy(model.state_dict()),
                "optimizer_state_dict": optimizer.state_dict(),
                "metrics": metrics,
                "best_val_loss": best_val_loss,
            },
            stage2_best_path,
        )
        status = "Best checkpoint saved"
    else:
        epochs_without_improvement += 1
        status = (
            f"No qualifying improvement "
            f"({epochs_without_improvement}/{STAGE2_PATIENCE})"
        )

    print(
        f"\nStage 2 Epoch {epoch}/{STAGE2_EPOCHS} | "
        f"Train loss: {train_loss:.4f} | "
        f"Train accuracy: {train_accuracy:.2%} | "
        f"Val loss: {val_loss:.4f} | "
        f"Val accuracy: {val_accuracy:.2%} | "
        f"Time: {elapsed:.1f}s | {status}\n"
    )

    if epochs_without_improvement >= STAGE2_PATIENCE:
        print("Early stopping: validation loss has stopped improving.")
        break

# Restore the best Stage 2 model
checkpoint = torch.load(
    stage2_best_path,
    map_location=device,
    weights_only=True,
)
model.load_state_dict(checkpoint["model_state_dict"])

print("\nSTAGE 2 COMPLETE")
print(f"Best epoch: {best_epoch}")
print(f"Best validation loss: {best_val_loss:.4f}")
print(f"Best checkpoint: {stage2_best_path}")
print(f"History: {stage2_history_path}")
```
